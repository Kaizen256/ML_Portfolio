Every thing I did to "improve" the baseline such as adding extra features or making a complex NN, has resulted in the WMAE score getting worse than the baseline resulting in a negative score. I am reverting back to my baseline and I am going to extensively test what improves the model, and what does not improve the model.

```python
import os
import re
import numpy as np
import pandas as pd
from pathlib import Path

from sklearn.decomposition import PCA, TruncatedSVD
from sklearn.linear_model import Ridge

import anndata as ad
import scanpy as sc
from scipy import sparse
import sys
```

```python
SEED = 6
EMBED_DIM = 128        # gene embedding size from control cells
OUT_PCA_K = 64         # number of output PCA components
RIDGE_ALPHA = 1.0      # ridge strength
CTRL_CELL_SAMPLE = 15000  # sample control cells for speed/memory
np.random.seed(SEED)
```

```python
from sklearn.model_selection import KFold
from myllia_metric import myllia_score

def score_delta(dt, dp):
    dt = dt.astype(np.float32, copy=False)
    dp = dp.astype(np.float32, copy=False)
    r = myllia_score(dt, dp)

    out = {}
    for k in ["score", "wcos", "mean_term", "pred_wmae"]:
        out[k] = float(getattr(r, k))
    return out

def cv_eval(predict_fn, genes, dt_true, n_splits=8, seed=6):
    genes = np.asarray(genes)
    dt_true = dt_true.astype(np.float32, copy=False)

    kf = KFold(n_splits=n_splits, shuffle=True, random_state=seed)
    fold = 0
    rows = []

    for tr_idx, va_idx in kf.split(genes):
        fold += 1
        genes_va = genes[va_idx].tolist()
        dt_va = dt_true[va_idx]

        dp_va = predict_fn(genes_va)
        s = score_delta(dt_va, dp_va)
        s["fold"] = fold
        s["n_val"] = int(len(va_idx))
        rows.append(s)

    scores = np.array([r["score"] for r in rows], dtype=float)
    return rows, float(scores.mean()), float(scores.std())
```

```python
df_means = pd.read_csv('data/training_data_means.csv')
df_valmap = pd.read_csv("data/pert_ids_val.csv")

gene_columns = [c for c in df_means.columns if c != "pert_symbol"]
baseline_mask = df_means["pert_symbol"] == "non-targeting"
x_base = df_means.loc[baseline_mask, gene_columns].iloc[0].to_numpy(np.float32)

df_train = df_means.loc[~baseline_mask].reset_index(drop=True)
train_genes = df_train["pert_symbol"].astype(str).values
missing_train = sorted({g for g in train_genes if g not in gene_columns})

X_train_means = df_train[gene_columns].to_numpy(dtype=np.float32) # Only gene columns
D_train = X_train_means - x_base[None, :] # Subtract baseline
n_train = D_train.shape[0] # Training perturbations

delta_baseline = D_train.mean(axis=0)
val_map = dict(zip(df_valmap["pert_id"].astype(str), df_valmap["pert"].astype(str)))

adata = ad.read_h5ad("Data/training_cells.h5ad")

ctrl_mask = adata.obs["sgrna_symbol"] == "non-targeting"
adata_ctrl = adata[ctrl_mask].copy()

sc.pp.normalize_total(adata_ctrl, target_sum=1e4, inplace=True)

varnames = adata_ctrl.var_names.astype(str)
gene_to_idx = {g: i for i, g in enumerate(varnames)}

missing_outputs = [g for g in gene_columns if g not in gene_to_idx]
if missing_outputs:
    print("[warn] output genes missing in h5ad:", missing_outputs[:20])

ordered_idx = [gene_to_idx[g] for g in gene_columns if g in gene_to_idx]
adata_ctrl = adata_ctrl[:, ordered_idx].copy()

X = adata_ctrl.X.tocsr(copy=True)
X.data = np.log2(X.data + 1.0).astype(np.float32)
```

```python
svd = TruncatedSVD(n_components=EMBED_DIM, random_state=SEED)
svd.fit(X)  # X: (n_cells, 5127)
```

```python
gene_emb = svd.components_.T.astype(np.float32)
gene2emb = {g: gene_emb[i] for i, g in enumerate(gene_columns)}
emb_fallback = gene_emb.mean(axis=0)
```

```python
out_pca = PCA(n_components=min(OUT_PCA_K, n_train - 1), random_state=SEED)
C_train = out_pca.fit_transform(D_train)  # (n_train, K)
K = C_train.shape[1]

X_feat = np.vstack([gene2emb.get(g, emb_fallback) for g in train_genes]).astype(np.float32)
reg = Ridge(alpha=RIDGE_ALPHA, random_state=SEED)
```

```python
reg.fit(X_feat, C_train)
```

```python
def soft_threshold(x: np.ndarray, t: float) -> np.ndarray:
    # elementwise: sign(x) * max(|x| - t, 0)
    return np.sign(x) * np.maximum(np.abs(x) - t, 0.0)

def predict_delta_for_gene(gene_symbol: str, scale: float = 1.0, thresh: float = 0.0) -> np.ndarray:
    e = gene2emb.get(gene_symbol, emb_fallback)[None, :]
    c_hat = reg.predict(e)  # (1, K)
    d_hat = out_pca.inverse_transform(c_hat).astype(np.float32)[0]  # (5127,)
    alpha = 0.2  # tune later
    d_hat = alpha * d_hat + (1 - alpha) * delta_baseline.astype(np.float32)
    if thresh > 0:
        d_hat = soft_threshold(d_hat, thresh).astype(np.float32)
    return d_hat
```

```python
def make_fold_predictor(tr_idx):
    # Fit output PCA and ridge ONLY on training fold
    D_tr = D_train[tr_idx]

    out_pca_fold = PCA(
        n_components=min(OUT_PCA_K, len(tr_idx) - 1),
        random_state=SEED
    )
    C_tr = out_pca_fold.fit_transform(D_tr)

    X_tr = np.vstack([gene2emb.get(g, emb_fallback) for g in train_genes[tr_idx]]).astype(np.float32)
    reg_fold = Ridge(alpha=RIDGE_ALPHA, random_state=SEED)
    reg_fold.fit(X_tr, C_tr)

    def predict_genes(genes_list, scale=1.0, thresh=0.0):
        Xv = np.vstack([gene2emb.get(g, emb_fallback) for g in genes_list]).astype(np.float32)
        C_hat = reg_fold.predict(Xv)
        D_hat = out_pca_fold.inverse_transform(C_hat).astype(np.float32)

        if scale != 1.0:
            D_hat *= np.float32(scale)

        if thresh > 0:
            D_hat = soft_threshold(D_hat, thresh).astype(np.float32)

        return D_hat

    return predict_genes
```

```python
from sklearn.model_selection import KFold

kf = KFold(n_splits=8, shuffle=True, random_state=SEED)
rows_cv = []

for fold, (tr_idx, va_idx) in enumerate(kf.split(train_genes), 1):
    pred_fn = make_fold_predictor(tr_idx)

    dp_va = pred_fn(train_genes[va_idx].tolist(), scale=1.0, thresh=0.0)
    dt_va = D_train[va_idx]

    s = score_delta(dt_va, dp_va)
    s["fold"] = fold
    s["n_val"] = int(len(va_idx))
    rows_cv.append(s)

df_cv = pd.DataFrame(rows_cv)
print(df_cv)
print(f"[cv score] mean={df_cv['score'].mean():.6f} std={df_cv['score'].std():.6f}")
```
